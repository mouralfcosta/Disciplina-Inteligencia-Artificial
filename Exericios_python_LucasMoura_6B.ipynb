21- Busca Binaria

```python
numeros = [1, 3 ,5, 7, 9, 11, 13, 15]
valor = int(input("Digite um valor: "))

inicio = 0
fim = len(numeros) - 1

while inicio <= fim:
    meio = (inicio + fim) // 2
    if numeros[meio] == valor:
        print(f"O valor {valor} está na posição {meio}.")
        break
    elif numeros[meio] < valor:
        inicio = meio + 1
    else:
        fim = meio - 1

else:
  print("Valor não encontrado ")
```

22- Criptografia - Cifra de Cesar

```python
def cifra_cesar(texto, deslocamento):
    resultado = ""
    for caractere in texto:
        if caractere.isalpha():
            inicio =ord('a') if caractere.islower() else ord('A')
            caractere_codificado = chr((ord(caractere) - inicio + deslocamento) % 26 + inicio)
            resultado += caractere_codificado
        else:
          resultado += caractere
    return resultado
mensagem = input("Digite uma mensagem: ")
deslocamento = int(input("Digite um deslocamento: "))
mensagem_codificada = cifra_cesar(mensagem, deslocamento)
print("Mensagem codificada:", mensagem_codificada)
```

24- Validação de CPF brasileiro

```python
def validar_cpf(cpf):
    if len(cpf) != 11:
        return False
    if cpf == cpf[0] * 11:
        return False

    soma= 0
    for n in range(9):
        soma += int(cpf[n]) * (10 - n)
    digito1 = (soma*10)%11
    if digito1==10:
        digito1 = 0
    if digito1 != int(cpf[9]):
      return False

    soma = 0
    for n in range(10):
        soma += int(cpf[n]) * (11 - n)
    digito2 = (soma*10)%11
    if digito2 == 10:
        digito2 = 0
    return digito2 == int(cpf[10])

cpf = input("Digite um CPF: ")

if validar_cpf(cpf):
  print("CPF válido")
else:
  print("CPF inválido")
```

40-Crie um programa para gerenciar uma lista de compras, permitindo
adicionar, remover e listar itens.

```python
lista = []

while True:
  print("1. Adicionar item")
  print("2. Remover item")
  print("3. Listar itens")
  print("4. Sair")

  opcao =input("Escolha uma opção: ")

  if opcao == "1":
    item = input("Digite o item a ser adicionado: ")
    lista.append(item)
    print("Item adicionado com sucesso")

  elif opcao =="2":
    item = input("Digite o item a ser removido: ")
    if item in lista:
      lista.remove(item)
      print("Item removido com sucesso")
    else:
      print("Item não encontrado na lista")

  elif opcao == "3":
    print("Itens na lista:")
    for item in lista:
      print(item)

  elif opcao == "4":
    print("Saindo do programa")
    break

  else:
    print("Opção invalida")
```

48-Implemente o algoritmo de ordenação por inserção (Insertion Sort).

```python
numeros = [5,4,7,1,9,6]

for i in range(1, len(numeros)):
  atual = numeros[i]
  j = i - 1
  while j >= 0 and numeros[j] > atual:
    numeros[j + 1] = numeros[j]
    j -= 1
    numeros[j + 1] = atual

print("numeros em ordenção : ",numeros)
```
