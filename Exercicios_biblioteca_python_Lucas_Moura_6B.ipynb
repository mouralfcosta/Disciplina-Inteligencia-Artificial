19- Crie dois arrays de numeros inteiros aleatorios de tamanho 20 e calcule o
angulo entre eles (produto escalar normalizado).

```python
import numpy as np

array1 = np.random.randint(1, 101, 20)
array2 = np.random.randint(1, 101, 20)

produto =np.dot(array1, array2) #vai calcular o produto escalar entre dois vetores
norma1 = np.linalg.norm(array1) #calcula o comprimento do array
norma2 = np.linalg.norm(array2)
angulo = np.arccos(produto / (norma1 * norma2)) #arco cosseno

angulo_graus = np.degrees(angulo)

print(array1)
print("-------")
print(array2)
print("-------")
print("angulo",angulo)
print("-------")
print("angulo graus",angulo_graus)
```

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.random.rand(100)
y = np.random.rand(100)

plt.scatter(x, y, c=y)# scatter cria um grafico de pontos e o c=y faz com que a cor dos pontos dependem do valor de y
plt.title("Gráfico de Dispersão")
plt.xlabel("Eixo X")
plt.ylabel("Eixo Y")
plt.show()
```

5- Faca um grafico de pizza para representar as porcentagens de quatro setores de mercado com rotulos personalizados

6- Plote dois graficos de linha em subplots diferentes: y1 = ex e y2 = log(x).

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(1, 10, 100)
y1 = np.exp(x)
y2 = np.log(x)

plt.subplot(2,1,1) # 2 linhas 1 coluna e 1 grafico
plt.plot(x, y1)#cria o grafico de linha
plt.title("y1 = e^x")

plt.subplot(2,1,2)
plt.plot(x, y2)
plt.title("y2 = log(x)")

plt.tight_layout() #cria um espaço entre os graficos
plt.xlabel("x")
plt.ylabel("y")
plt.show()
```

8-Plote um grafico de linha com intervalos de confianca (sombramento) para
a funcao y = sin(x).

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 10, 100)
y = np.sin(x)

plt.plot(x, y)
plt.fill_between(x, y-0.1, y+0.1, alpha=0.3) #area entre duas curvas
plt.title("Gráfico de Linha com Intervalos de Confiança")
plt.xlabel("Eixo X")
plt.ylabel("Eixo Y")
plt.show()
```

10-Plote tres graficos de linha em um unico grafico, cada um com uma cor e
estilo de linha diferentes

```python
from matplotlib import lines
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 10, 100)

plt.plot(x, x, color = "green", linestyle = "solid")#solido
plt.plot(x, x**2, color = "blue", linestyle = "dashed")#tracejada
plt.plot(x, x**3, color = "red", linestyle = "dotted") #pontilhada

plt.legend(["Linha 1", "Linha 2", "Linha 3"])# cria a legenda

plt.title("Gráfico de Linhas com Cores e Estilos")
plt.xlabel("Eixo X")
plt.ylabel("Eixo Y")
plt.show()
```

11-Crie um grafico de barras empilhadas para tres categorias com tres grupos
de valores.

```python
import matplotlib.pyplot as plt
import numpy as np

categorias = ["A", "B", "C"]
grupo1 = [10, 20, 30]
grupo2 = [20, 30, 40]
grupo3 = [30, 40, 50]

plt.bar(categorias, grupo1, label="Grupo 1")
plt.bar(categorias, grupo2, bottom=grupo1, label="Grupo 2") #bottom #faz com que onde a proxima barra começa
plt.bar(categorias, grupo3, bottom=np.array(grupo1)+np.array(grupo2), label="Grupo 3")

plt.title("Grafico de barras empilhadas")
plt.xlabel("Categorias")
plt.ylabel("Valores")
plt.legend()
plt.show()
```

12-Plote um grafico de linha com uma funcao quadratica e ajuste os limites
do eixo para focar nos valores x = −5 a x = 5

```python
import matplotlib.pyplot as plt
import numpy as np

x=np.linspace(-10, 0, 100)
y=x**2
plt.plot(x, y)

plt.xlim(-5, 5)#limite do x
plt.ylim(0,30)#limite do y

plt.title("Gráfico de Linha com Função Quadrática")
plt.xlabel("Eixo X")
plt.ylabel("Eixo Y")
plt.show()
```

13-Gere um grafico de dispersao 3D para 100 pontos aleatorios utilizando
Matplotlib.

```python
import matplotlib.pyplot as plt
import numpy as np
from mpl_toolkits.mplot3d import Axes3D

x = np.random.rand(100)
y = np.random.rand(100)
z = np.random.rand(100)

fig = plt.figure()#cria a figura
ax = fig.add_subplot(111, projection="3d") #cria a projeção em 3d

ax.scatter(x, y, z)

plt.title("Gráfico de Dispersão 3D")

plt.show()
```

19- Plote um grafico de linha com duas series e adicione uma legenda para
diferencia-las

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(1, 10, 100)
y1 = np.sin(x)
y2 = np.cos(x)

plt.plot(x, y1, label="seno")
plt.plot(x, y2, label="cosseno")

plt.title("Gráfico de Linha com Legenda")
plt.xlabel("Eixo X")
plt.ylabel("Eixo Y")
plt.legend()
plt.show()
```

20-Adicione uma grade com espacamento personalizado a um grafico de linha
